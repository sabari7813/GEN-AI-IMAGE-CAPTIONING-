```python

import torch
import gradio as gr

from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python

model_name = "Salesforce/blip-image-captioning-base"

processor = BlipProcessor.from_pretrained(model_name)

model = BlipForConditionalGeneration.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"

model.to(device)

print("Image Captioning Model Loaded Successfully!")
```

```python

def generate_caption(image):

    if image is None:
        return "Please upload an image."

    inputs = processor(
        image.convert("RGB"),
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():

        output = model.generate(
            **inputs,
            max_new_tokens=50
        )

    caption = processor.decode(
        output[0],
        skip_special_tokens=True
    )

    return caption
```

```python

demo = gr.Interface(
    fn=generate_caption,
    inputs=gr.Image(type="pil", label="Upload Your Image"),
    outputs=gr.Textbox(label="Generated Image Caption"),
    title="Gen AI Powered Image Captioning",
    description="Upload an image and let AI generate a meaningful caption."
)
```

```python

demo.launch()
```
